```python
import pandas as pd
import numpy as np

INPUT_PATH = "../data/raw/Loan_default.csv"
OUTPUT_PATH = "../data/processed/loan_default_clean.csv"

df = pd.read_csv(INPUT_PATH)

print(df.shape)
```

```python
df = df.drop(columns=["LoanID"])
```

```python
print("Duplicate rows:", df.duplicated().sum())

#df = df.drop_duplicates()
```

```python
df.isnull().sum()

#df= df.dropna
```

```python
numeric_columns = [
    "Age",
    "Income",
    "LoanAmount",
    "CreditScore",
    "MonthsEmployed",
    "NumCreditLines",
    "InterestRate",
    "LoanTerm",
    "DTIRatio"
]
categorical_columns = [
    "Education",
    "EmploymentType",
    "MaritalStatus",
    "HasMortgage",
    "HasDependents",
    "LoanPurpose",
    "HasCoSigner"
]
df[numeric_columns].describe().T
```

```python
print(df[df["Age"] < 18])
print(df[df["Income"] <= 0])
print(df[df["LoanAmount"] <= 0])
print(df[df["CreditScore"] < 300])
print(df[df["CreditScore"] > 850])
print(df[df["DTIRatio"] < 0])
```

```python
for column in categorical_columns:
    print(column)
    print(df[column].unique())
```

```python
#For white space
for column in categorical_columns:
    if df[column].dtype == "object":
        df[column] = df[column].str.strip()
```

```python
#For target
print(df["Default"].value_counts())
print(df["Default"].unique())
```

```python
from pathlib import Path

output_dir = Path("../data/processed")
output_dir.mkdir(parents=True, exist_ok=True)

OUTPUT_PATH = output_dir / "loan_default_clean.csv"

df.to_csv(OUTPUT_PATH, index=False)

print("Saved:", OUTPUT_PATH)
print("Final shape:", df.shape)
```
