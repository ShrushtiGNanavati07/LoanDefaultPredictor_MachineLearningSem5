```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("../data/processed/loan_default_clean.csv")
```

```python
sns.countplot(data=df, x="Default")

plt.title("Loan Default Distribution")
plt.show()
```

```python
sns.boxplot(
    data=df,
    x="Default",
    y="Income"
)

plt.title("Income vs Loan Default")
plt.show()
```

```python
sns.boxplot(
    data=df,
    x="Default",
    y="CreditScore"
)

plt.title("Credit Score vs Loan Default")
plt.show()
```

```python
sns.boxplot(
    data=df,
    x="Default",
    y="LoanAmount"
)

plt.title("Loan Amount vs Loan Default")
plt.show()
```

```python
sns.boxplot(
    data=df,
    x="Default",
    y="DTIRatio"
)

plt.title("Debt-to-Income Ratio vs Loan Default")
plt.show()
```

```python
sns.countplot(
    data=df,
    x="Education",
    hue="Default"
)

plt.xticks(rotation=30)
plt.title("Education Level vs Default")
plt.show()
```

```python
sns.countplot(
    data=df,
    x="EmploymentType",
    hue="Default"
)

plt.xticks(rotation=30)
plt.title("Employment Type vs Default")
plt.show()
```

```python
import numpy as np
numeric_df = df.select_dtypes(include=np.number)

plt.figure(figsize=(12, 8))

sns.heatmap(
    numeric_df.corr(),
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Numerical Feature Correlation")
plt.show()
```
